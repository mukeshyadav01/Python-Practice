#Operator in Python

```python
#arithmetic Operator
print(3+3)
print(5-2)
print(3*3)
print(8%2)
print(5**2)
print(5//2)
```

```python
# Relation operator
print(4<5)

print(4>5)

print(4<=5)

print(4==5)

print(4!=4)
```

```python
#Logical Operator
print(1 and 0)
print( 1 or 0)
print (not 1)
```

```python
#Bitwise Operator { it is a special operator run on binary}

#bitwise and
print( 2 & 3)

# Bitwise or
print (2 | 3)

# Bitwise xor
print(2 ^ 3)

print(~3)
print(4 >> 2)
```

```python
# Assignment Operator
# = is used to assign
a=2
# a = a %2
a +=2
# a++ ++a is not used in Python
print(a)
```

```python
# Membership Operator { ye batata hai ki koi chiz usame hai ya nahi }
# in/not in
print('d' in 'delhi')
print ( 1 in [2,3,5])
```

```python
# Program : Find the sum of a 3 digit number entered by the user
number = int(input("Enter the Number: "))
a = number%10
number = number//10
b = number % 10
number = number //10
c = number%10
print(a + b + c)
```

```python
# using while loop ,the program for table
number = int(input ('enter the number:'))
i = 1
while i<11:
  print(number,'*',i,'=', number*i)
  i = i+1
```

```python
# while loop with else
x=1
while x<3:
  print(x)
  x +=1
else:
  print('limit crossed')
```

```python
#gussing game
import random

jackpot = random.randint(1, 100)

guess = int(input("Guess the number: "))
counter = 1

while guess != jackpot:

    if guess < jackpot:
        print("Galat hai, thoda bada socho")
    else:
        print("Galat hai, thoda niche socho")

    counter = counter + 1
    guess = int(input("Guess again: "))
else:
  print("Correct!")
  print("The number of attempts:", counter)
```
